```python
from pyspark.sql import functions as F

# Read Silver table
df_silver = spark.read.table("zomato_silver")

# Gold Layer — Top 10 locations by average rating
df_gold_locations = df_silver \
    .groupBy("location") \
    .agg(
        F.round(F.avg("rate"), 2).alias("avg_rating"),
        F.count("name").alias("total_restaurants"),
        F.round(F.avg("approx_cost"), 0).alias("avg_cost")
    ) \
    .orderBy(F.desc("avg_rating")) \
    .limit(10)

display(df_gold_locations)
```

```python
# Save Gold layer
df_gold_locations.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("zomato_gold_locations")

print("Gold Delta Table saved! ")
```

```python
# Most popular cuisines in Bangalore
df_gold_cuisines = df_silver \
    .groupBy("cuisines") \
    .agg(
        F.count("name").alias("restaurant_count"),
        F.round(F.avg("rate"), 2).alias("avg_rating"),
        F.round(F.avg("approx_cost"), 0).alias("avg_cost")
    ) \
    .orderBy(F.desc("restaurant_count")) \
    .limit(10)

display(df_gold_cuisines)
```

```python
# Save Gold cuisines
df_gold_cuisines.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("zomato_gold_cuisines")

print("Gold Cuisines Table saved! ")
```
