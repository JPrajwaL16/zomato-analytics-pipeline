```python
# Download Zomato Bangalore dataset
import kaggle

kaggle.api.dataset_download_files(
    'himanshupoddar/zomato-bangalore-restaurants',
    path='/tmp/zomato_data',
    unzip=True
)

print("Dataset Downloaded Successfully!")
```

```python
# Check downloaded files
import os

files = os.listdir('/tmp/zomato_data')
print("Files downloaded:")
for file in files:
    print(file)
```

```python
# Move file from /tmp to Workspace
import shutil
import os

# Create destination folder
os.makedirs("/Workspace/swiggy_analytics_project/data", exist_ok=True)

# Move the file
shutil.copy(
    '/tmp/zomato_data/zomato.csv',
    '/Workspace/swiggy_analytics_project/data/zomato.csv'
)

print("File moved successfully!")
```

```python
# Now read from Workspace path
df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("/Workspace/swiggy_analytics_project/data/zomato.csv")

print(f"Total Rows: {df.count()}")
print(f"Total Columns: {len(df.columns)}")
```
